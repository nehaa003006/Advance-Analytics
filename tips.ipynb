```python
#Add a new column
df['tip_percentage'] = 100*df['tip']/df['total_bill']
df['price_per_person']=df['total_bill']/df['size']
print(df.head())
df['price_per_person']=np.round(df['price_per_person'],2)
print(df.head())
```

```python
#Drop a column
df=df.drop("tip_percentage", axis=1) #axis=1 means drop a column
#df.drop("tip_percentage",axis=1,inplace=True) #Another syntax
print(df.head())
```

```python
#Set a column as index - See how the display looks like
df=df.set_index('Payment ID')
print(df.head())
df=df.reset_index()
print(df.head())
```

```python
#Conditional filtering
print(df['total_bill']<30) #Boolean Series(True/False)
print(df[df['total_bill'] <30]) #Actual rows where total_bill < 30
print(df[df['gender']=='Male'])
print(df[(df['total_bill'] > 
          30) & (df['gender'] =='Male')]) 
print(df[(df['total_bill'] > 
          30) & ~(df['gender'] =='Male')]) # ~ is negation
print(df[(df['total_bill']>30) | (df['tip']>5)])
```

```python
options = ['Sat', 'Sun']
print(df[df['day'].isin(options)])
```
