```python
#Consider Ho= Doing PG-DBDA course does not improve the chance of palacements and they remain at 40% We run 4 test- Consider x bar =(1) 80,(2) 41, (3) 45 ,(4) 35 consider n =100
```

```python
import math
from scipy.stats import norm
```

```python
#Given 
mu=40 #(hypothised mean 40%)
sample_means=[80,41,45,-35]
n=100
s=15
alpha=0.05
```

```python
#critical values
#norm.cdf() gives probability , norm.ppf() gives z score
#ppf=percent point function(Inverse of CDF)
#Gives z score corresponding to a given probability
#eg tell me the Zscore such that the area under the curve (left of it is p)
z_crit_two_tailed=norm.ppf(1-alpha/2) #1.96
z_crit_one_tailed=norm.ppf(1-alpha) #1.645
```

```python
print("Critical values")
print("-----------")
print(f"Two tailed(alpha=0.05):+-{z_crit_two_tailed:.3f}")
print(f"One tailed(alpha=0.05):+-{z_crit_one_tailed:.3f}")
print("\n")
```

```python
#Function to compute z
def compute_z(xbar):
    return(xbar-mu)/(s/math.sqrt(n))
    
```

```python
#Evaluation
for xbar in sample_means:
    print(f"=====For sample mean x bar ={xbar}=====")
    z=compute_z(xbar)
```

```python
#p-values
#norm.cdf(z) gives the probability to the LEFT of z
#SO 1 norm.cdf(z) gives the probability to the right and double
#e.g Z = 2 ,Right-tail area =0..228
#So total area= 2 x 0.0228=0.0456
#For one-tailed:one-sided only
#p_two_tailed=2*(1-norm.cdf(abs(z)))
p_two_tailed=2*norm.sf(abs(z))
p_one_tailed=1- norm.cdf(z) #improvement only (right-tail)
```

```python
print(f"Zscore:{z:.3f}")
```

```python
#Two tailed decision
if abs(z)>z_crit_two_tailed:
    two_tailed_decision="Reject HO (Significant difference)"
else:
    two_tailed_decision="Fail to Reject HO (No Significant difference)"
```

```python
#One-tailed decision
if z>z_crit_one_tailed:
    one_tailed_decision="Reject HO (Significant improvement)"
else:
    one_tailed_decision="Fail to Reject HO (No Significant improvement)"
```

```python
print(f"Two-tailed test")
print(f"p-value={p_two_tailed:.4f}")
print(f"Decision:{two_tailed_decision}")
```

```python
print(f"One-tailed test")
print(f"p-value={p_one_tailed:.4f}")
print(f"Decision:{one_tailed_decision}")
```
