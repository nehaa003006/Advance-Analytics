```python
#LPP SOLUTION USING PuLP
from pulp import *
```

```python
#Create the problem variable to contain the problem data
model=LpProblem("FactorypProductionProblem",LpMaximize)
```

```python
#Create variables for the chairs and the table
#pARAMETERS : nMAE , lOWER IMIT , upper limit, Data Type
chairs=LpVariable("Chairs",0,None,LpInteger) # Number of chairs to produce
tables=LpVariable("Tables",0,None,LpInteger) # Number of tables to produce
```

```python
#Create maximize objective function
model+=300*chairs+1000*tables,"Profit"
```

```python
#create constraints
model+=5*chairs+10*tables<=400,"Wood Constraint" #Wood availability constraint
model+=2*chairs+6*tables<=300,"Time Constraint" #worker time constraint
```

```python
# the problem is solved using the PuLP choice of Solver
model.solve()
```

```python
#Each of the variables is printed with its resolved optimum value
for v in model.variables():
    print(v.name,'=',v.varValue)
```

```python
#Additional Information
print("Total Profit",value(model.objective)) 
#Print objective function value
```

## Car Problem

```python
#create the problem variable
model1=LpProblem("MaximizeTheCarProfit",LpMaximize)
```

```python
#creating variables for the number of car A and car B
car_A=LpVariable("A",0,None,LpInteger) # the 0 signifizes the non negative condition and the None is no limit on the upper condition
car_B=LpVariable("B",0,None,LpInteger)
```

```python
# create the maximize objective function
model1+=30000*car_A+450000*car_B,"Profit"
```

```python
#creating the constraints
model1+=3*car_A+4*car_B<=30,"Robot" #constraint 1
model1+=5*car_A+6*car_B<=60,"Engineer Constraint"
model1+=1.5*car_A+3*car_B<=21,"Detailer Constraint"
```

```python
model1.solve()
```

```python
for v in model1.variables():
    print(v.name,'=',v.varValue)
```

```python
print("Total Profit",value(model1.objective)) 
```

```python
#creating the problem variable
model2=LpProblem("ToGetTheMaximumCredit",LpMaximize)
```

```python
#creating the variables
subA=LpVariable("A",1,3,LpInteger)
subB=LpVariable("B",1,2,LpInteger)
subC=LpVariable("C",1,5,LpInteger)
```

```python
#creating the objective function
model2+=subA+subB+subC,"Total_credits"
```

```python
model2+=15*subA+12*subB+18*subC<=120,"Total_Study_Hours"
```

```python
model2.solve()
```

```python
for v in model2.variables():
    print(v.name,'=',v.varValue)
```

```python
print(value(model2.objective))
```

```python
emi_bank_a=2076
emi_bank_b=3227
emi_bank_c=4707
total_loan_required=10
```

```python
model3=LpProblem("LoanDistributionProblem",LpMinimize)
```

```python
x1=LpVariable("Loan_from_bankA",0,None,LpContinuous)
x2=LpVariable("Loan_from_bankB",0,None,LpContinuous)
x3=LpVariable("Loan_from_bankC",0,None,LpContinuous)
```

```python
model3+=emi_bank_a*x1+emi_bank_b*x2+emi_bank_c*x3,"Total_Monthly_EMI"
model3+=x1+x2+x3==total_loan_required,"Total_Loan_Constraint"
model3+=x1>=2,"Min_Loan_Bank_A"
model3+=x3<=4,"Min_Loan_Bank_C"
```

```python
model3.solve()
```

```python
for v in model3.variables():
    print(v.name,'=',v.varValue)
```
