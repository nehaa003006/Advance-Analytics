```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df= pd.read_csv("billionaires.csv")
```

```python
finalworth=df['finalWorth']
```

```python
mean_finalworth=finalworth.mean()
median_finalworth=finalworth.median()
skewness_finalworth=finalworth.skew()
kurtosis_finalworth=finalworth.kurt()
```

```python
print(mean_finalworth,median_finalworth,skewness_finalworth,kurtosis_finalworth)
```

```python
plt.figure(figsize=(8,6))
sns.histplot(finalworth,kde=True,color='skyblue',bins=30)
plt.axvline(mean_finalworth,color='red',linestyle='--', label=f'Median:{median_finalworth:.2f}')
plt.title(f'Distribution of Final Worth\n skewness:{skewness_finalworth:.2f},kurtosis:{kurtosis_finalworth:.2f}')
plt.xlabel('Final Worth')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(finalworth)
plt.tight_layout()
plt.show()
```

```python
#removing the wxtreme outliers and retry
Q1=df['finalWorth'].quantile(0.25)
Q3=df['finalWorth'].quantile(0.75)
IQR=Q3-Q1
lower_fence = Q1 -(3*IQR)
upper_fence = Q3+(3*IQR)
```

```python
df['is_extreme_outlier']=(df['finalWorth']>upper_fence)|(df['finalWorth']<lower_fence)
```

```python
num_extreme_outliers=df['is_extreme_outlier'].sum()
print(f" The Number of extreme outliers identified:{num_extreme_outliers}")
```

```python
#remove extreme outliers
df1=df[df['is_extreme_outlier']==False].copy()
finalworth=df1['finalWorth']
mean_finalworth=finalworth.mean()
median_finalworth=finalworth.median()
skewness_finalworth=finalworth.skew()
kurtosis_finalworth=finalworth.kurt()
print(skewness_finalworth)
```

```python
plt.figure(figsize=(8,6))
sns.histplot(finalworth,kde=True,color='skyblue',bins=30)
plt.axvline(mean_finalworth,color='red',linestyle='--', label=f'Mean:{mean_finalworth:.2f}')
plt.axvline(median_finalworth,color='black',linestyle='--', label=f'Median:{median_finalworth:.2f}')
plt.title(f'Distribution of Final Worth\n skewness:{skewness_finalworth:.2f},kurtosis:{kurtosis_finalworth:.2f}')
plt.xlabel('Final Worth')
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.boxplot(finalworth)
plt.tight_layout()
plt.show()
```

```python
#Compare CoV for two categories
group1_name='Fashion& Retail'
group2_name='Technology'
```

```python
group1=df[df['category']==group1_name]['finalWorth']
group2=df[df['category']==group2_name]['finalWorth']
```

```python
#Calculate CoV
cov_group1= group1.std()/group1.mean()*100
cov_group2= group2.std()/group2.mean()*100
```

```python
print(f" Coefficient of variation for {group1_name}:{cov_group1:.2f}%")
print(f" Coefficient of variation for {group2_name}:{cov_group2:.2f}%")
```
