```python
import numpy as np
```

```python
arr = np.arange(0,10)
```

```python
#Basic Arithmetic
print(arr + arr)
print(arr * arr)
print(arr-arr)
print(arr / arr) #division by zero (0/0=nan)
print(arr ** 3)
print(np.sqrt(arr))
```

```python
#Summary Statistics
print(arr.sum())
print(arr.mean())
print(arr.max())
print(arr.min())
print(arr.var())
print(arr.std())
```

```python
#Reversing an array
arr = np.array([1, 2, 3, 4, ])
reversed_arr = arr[::-1]
print(reversed_arr)
```

```python
#Sorting an array
arr = np.array([5, 2, 8, 1 , 3])
sorted_arr = np.sort(arr)
print(sorted_arr)
```

```python
#For Descending
sorted_arr = np.sort(arr[::-1]) #Reverse the array
print(sorted_arr)
```

```python
#Creating multi-dimensional arrays as a "structure"

students = np.array([
    ('Alice', 95, 20),
    ('Bob', 88, 21),
    ('Charlie', 92, 19)],dtype=[('name', 'U10'),('marks', 'i4'), ('age', 'i4')])

print(students)
print(students['name'])
print(students['marks'])
print(students['marks'].mean())
print(students[students['name'] == 'Alice']['marks'])
print(students[students['name'] == 'Alice']['marks'][0])
```

```python
sales = [0, 5, 155, 0, 518, 616]
sales_array = np.array(sales)
```

```python
print(type(sales_array))
print(sales_array.ndim)
print(sales_array.shape)
print(sales_array.size)
print(sales_array.dtype)
```

```python
#where() function
result_array = np.where(sales_array ==0, "No Sales", "Some sales")
print(result_array)
```

```python
#Show all boolean values where condition is applied
bool_mask = sales_array!= 0
print(bool_mask)
```

```python
#Filter using the boolean mask
filtered_sales = sales_array[bool_mask]
print(filtered_sales)
```
