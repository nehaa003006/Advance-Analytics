```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import zscore
```

```python
df=pd.read_csv("Diabetes.csv")
```

```python
df['Glucose']
```

```python
glucose_mean=df['Glucose'].mean()
glucose_mean
```

```python
df['glucose_zscore']=zscore(df['Glucose'])
```

```python
zscore_threshold=3
```

```python
glucose_outlier= df[df['glucose_zscore'].abs()>zscore_threshold]
```

```python
print(f'Number of glucose outlier:{len(glucose_outlier)}')
```

```python
plt.figure(figsize=(14,6))
#Plot for weight
plt.subplot(1,2,1) #1 row, 2 columns, 1st plot
sns.histplot(df['Glucose'], kde=True,color='skyblue')
plt.scatter(glucose_outlier['Glucose'],[0]*len(glucose_outlier),color='red',s=50,zorder=5,label=f'Outliers(z>{zscore_threshold})')
plt.axvline(glucose_mean,color='green',linestyle='--',label=f'Mean:{glucose_mean:.2f}')
plt.title('Distribution of Glucose with outliers')
plt.xlabel('Glucose')
plt.ylabel('Glucose')
plt.legend()
plt.grid(True,linestyle='--',alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
#Box Plot for Weight
plt.figure(figsize=(6,8))
sns.boxplot(y=df['Glucose'],color='lightgreen')
plt.title('Box PLot for Glucose')
plt.ylabel('Glucose')
plt.grid(True,linestyle='--',alpha=0.7)
plt.show()
```
