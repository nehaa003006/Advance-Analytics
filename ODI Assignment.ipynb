```python
import pandas as pd 
import numpy as np
from scipy.stats import binom
```

```python
Match_20 = pd.read_csv('Men ODI Team Match Results - 20th Century.csv')
Match_21 = pd.read_csv('Men ODI Team Match Results - 21st Century.csv')
df=pd.concat([Match_20, Match_21])
```

```python
df= df[df['Result'].isin(['Won','Lost','Tied'])]
```

```python
indian_matches = len(df[df['Country'].str.contains('India')])
indian_matches
```

```python
India_win = len(df[(df['Country'].str.contains('India'))&(df['Result'].isin(['Won']))])
India_win
```

```python
india_win_rate=India_win/indian_matches
india_win_rate
```

```python
total=60
```

```python
p_exact= binom.pmf(35,60,india_win_rate)
p_exact
```

```python
p_more_than_40 = 1 - binom.cdf(40,60,india_win_rate)
p_more_than_40
```

```python
p_at_least_40 = 1 - binom.cdf(44,60,india_win_rate)
p_at_least_40
```

```python
matches = pd.Series(np.arange(1,60))
pmf_list = matches.apply(lambda x: binom.pmf(x, total, india_win_rate))
max_prob_win = matches.iloc[pmf_list.idxmax()]
```

```python
print(p_exact,p_more_than_40,p_at_least_40)
```
