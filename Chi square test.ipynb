```python
#chi square test of independece,also first create the contingency table  using the crosstab() function
#HO there is no dependence of passanger class on the passanger survival
import pandas as pd
from scipy.stats import chi2_contingency
```

```python
df=pd.read_csv("titanic.csv")
```

```python
df.columns
```

```python
#Create a contingency table
contingency_table=pd.crosstab(df['pclass'],df['survived'])
print(contingency_table)
```

```python
#perform the chi square test
chi2_stat,p_value,dof,expected=chi2_contingency(contingency_table)
```

```python
print("Chi-square statistics", chi2_stat)
print("p-value",p_value)
print("Degrees of freedom",dof)
```

```python
if p_value<0.05:
    print("Reject HO")
else:
    print("Do not reject HO")
```
