# Stratified random sampling

```python
total_amer_indian_eskimo = len(df[df['race']=='Amer-Indian-Eskimo'])
total_asian_pac_islander = len(df[df['race']=='Asian-Pac-Islander'])
total_black = len(df[df['race']=='Black'])
total_white = len(df[df['race']=='White'])
total_other = len(df[df['race']=='Other'])
```

```python
sample_size = int(0.1 * original_row_count)
sample_amer_indian_eskimo = int(0.1 * total_amer_indian_eskimo)
sample_asian_pac_islander = int(0.1 * total_asian_pac_islander)
sample_black = int(0.1 * total_black)
sample_white = int(0.1 * total_white)
sample_other = int(0.1 * total_other)
print(sample_size,sample_amer_indian_eskimo,sample_asian_pac_islander,sample_black,sample_white,sample_other)
```

```python
#Generating proportionate samples
amer_indian_eskimo_sample=df[df['race'] =='Amer-Indian-Eskimo'].sample(n=sample_amer_indian_eskimo,random_state=42)
asian_pac_islander_sample=df[df['race'] =='Asian-Pac-Islander'].sample(n=sample_asian_pac_islander,random_state=42)
black_sample=df[df['race'] =='Black'].sample(n=sample_black,random_state=42)
white_sample=df[df['race'] =='White'].sample(n=sample_white,random_state=42)
other_sample=df[df['race'] =='Other'].sample(n=sample_other,random_state=42)
print(len(amer_indian_eskimo_sample),len(asian_pac_islander_sample),len(black_sample),len(white_sample),len(other_sample))
```

```python
#Combining into one single dataframe
sample_df=pd.concat([amer_indian_eskimo_sample,amer_indian_eskimo_sample,black_sample,white_sample,other_sample])
sample_row_count=len(sample_df)
print(df.head())
print(sample_df.head())
```

## Cluster Sampling

```python
df.value_counts('marital.status')
```

```python
marital_status=df['marital.status'].unique()
num_marital_status=len(marital_status)
print(marital_status,num_marital_status)
np.random.seed(42)
selected_marital_status =np.random.choice(marital_status,size=num_marital_status//2)
sample_df_4=df[df['marital.status'].isin(selected_marital_status)]
sample_df_4
```

```python
sample_df_5 = sample_df_4.groupby('marital.status')
sample_df_5=sample_df_5.apply(lambda x:x.sample(frac=0.278, random_state = 42),include_groups=False).reset_index(drop=True)
sample_df_5
```

```python
clustered_avg = sample_df_5['education.num'].mean()
clustered_avg
```
