```python
import pandas as pd
```

```python
#lINEAR Regression
df=pd.read_csv('salary.csv')
X=df[['YearsExperience']]
y=df[['Salary']]
```

```python
#Step 2 : Split the dataset into trainig and testing
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(X, y,test_size=0.20,random_state=0)
```

```python
#Step 3: Fit Simple Linear Regression to Training data
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
#Step 4: Make Prediction
y_pred=model.predict(X_test)
print(y_pred)
```

```python
#Step5: Visualize training set results
import matplotlib.pyplot as plt
plt.scatter(X_train,y_train,color='red')
plt.plot(X_train,model.predict(X_train),color='blue')
plt.title('Salary vs Experience(Training set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
#Step 6 - Visualize test set result
#Repeat the above, only change 'train' to test
plt.scatter(X_test,y_test,color='red')
plt.plot(X_test,model.predict(X_test),color='blue')
plt.title('Salary vs Experience(Test set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

```python
#Step 7 -Make the prediction
new_X_values_df= pd.DataFrame({'YearsExperience':[1,3,5,8,15,20,25]})
new_y_pred=model.predict(new_X_values_df)
print(new_y_pred)
```

```python
#Step 8 -Model Evaluation
from sklearn.metrics import mean_absolute_error, mean_squared_error,r2_score
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
r2 = r2_score(y_test,y_pred)
print(mae,mse,r2)
```

Assignment 3 Predictive weight using in weight-height .csv dataset


```python
df=pd.read_csv('weight-height.csv')
X=df[['Height']]
y=df[['Weight']]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test=train_test_split(X, y,test_size=0.20,random_state=0)
```

```python
#Step 3: Fit Simple Linear Regression to Training data
from sklearn.linear_model import LinearRegression
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
#Step 4: Make Prediction
y_pred=model.predict(X_test)
print(y_pred)
```

```python
#Step5: Visualize training set results
import matplotlib.pyplot as plt
plt.scatter(X_train,y_train,color='skyblue')
plt.plot(X_train,model.predict(X_train),color='blue')
plt.title('Weight vs height(Training set)')
plt.xlabel('height')
plt.ylabel('weight')
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.scatter(X_test,y_test,color='skyblue')
plt.plot(X_test,model.predict(X_test),color='blue')
plt.title('Weight vs height(Test set)')
plt.xlabel('height')
plt.ylabel('weight')
plt.show()
```

```python
#Step 7 -Make the prediction
new_X_values_df= pd.DataFrame({'Height':[20,34,29,35,34,20,25]})
new_y_pred=model.predict(new_X_values_df)
print(new_y_pred)
```
